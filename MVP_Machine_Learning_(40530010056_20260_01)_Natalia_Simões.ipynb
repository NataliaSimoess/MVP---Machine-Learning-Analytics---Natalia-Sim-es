##4.4 Análise das variáveis categóricas

```python
# Distribuição das principais variáveis categóricas

variaveis = ["gender", "region", "coverage_level"]

plt.figure(figsize=(18,5))

for i, coluna in enumerate(variaveis, 1):

    percentuais = (
        df[coluna]
        .value_counts(normalize=True)
        .sort_index()
        .mul(100)
    )

    ax = plt.subplot(1, 3, i)

    percentuais.plot(
        kind="bar",
        ax=ax
    )

    for j, valor in enumerate(percentuais):
        ax.text(
            j,
            valor + 0.3,
            f"{valor:.1f}%",
            ha="center",
            fontsize=10
        )

    plt.title(f"Distribuição de {coluna}")
    plt.xlabel("")
    plt.ylabel("Percentual (%)")
    plt.ylim(0, max(percentuais) + 10)
    plt.xticks(rotation=25)

plt.tight_layout()
plt.show()
```

A distribuição das variáveis categóricas foi analisada com o objetivo de identificar possíveis desbalanceamentos que pudessem impactar o treinamento dos modelos de Machine Learning.
```python
# Criando faixas de idade
df["faixa_idade"] = pd.cut(
    df["age"],
    bins=[17, 30, 45, 55, 65],
    labels=["18-30", "31-45", "46-55", "56-65"]
)

plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="faixa_idade",
    y="charges"
)

plt.title("Custo do seguro por faixa etária")
plt.xlabel("Faixa etária")
plt.ylabel("Charges")

plt.show()
```

## 5.1 Separação entre X e y

```python
# Separação das variáveis preditoras (X) e da variável alvo (y)

X = df.drop(
    ["charges", "bmi_categoria", "faixa_idade"],
    axis=1
)

y = df["charges"]

# Verificação das dimensões
print(f"Variáveis preditoras (X): {X.shape}")
print(f"Variável alvo (y): {y.shape}")
```

```python
print("Primeiras linhas das variáveis preditoras:")
display(X.head())

print("\nPrimeiros valores da variável alvo:")
display(y.head())
```

Nesta etapa, a variável `charges` foi separada das demais por representar a variável alvo do problema de regressão, ou seja, o valor que os modelos deverão aprender a prever.

Além da remoção da variável alvo, os atributos `bmi_categoria` e `faixa_idade`, criados exclusivamente para apoiar a análise exploratória dos dados, também foram excluídos da base de modelagem. Após essa etapa, o conjunto de dados passou a ser composto por 11 variáveis preditoras, que serão utilizadas para identificar padrões relacionados ao custo do seguro de saúde.

Essa separação evita vazamento de dados (*data leakage*), garantindo que o modelo não tenha acesso antecipado ao valor que deverá prever e permitindo uma avaliação mais confiável do desempenho preditivo.

## 5.2 Tratamento dos valores ausentes

Conforme identificado na análise exploratória, apenas as variáveis medical_history e family_medical_history apresentam valores ausentes.

Por se tratarem de variáveis categóricas, será utilizada a estratégia de substituição pela categoria mais frequente (most frequent), preservando a maior parte da informação disponível e evitando a perda de registros.

Esse tratamento será incorporado ao pipeline de pré-processamento, garantindo que a imputação seja realizada apenas com base nos dados de treinamento e posteriormente aplicada aos dados de teste, evitando vazamento de informações.

A exclusão dos registros com valores ausentes não foi adotada, pois aproximadamente 25% dos valores das variáveis `medical_history` e `family_medical_history` encontram-se ausentes. A remoção dessas observações resultaria em perda significativa de informações e redução do conjunto de treinamento, podendo impactar negativamente o desempenho dos modelos.

## 5.3 Construção do pipeline de pré-processamento

Os tratamentos definidos anteriormente foram reunidos em um único pipeline de pré-processamento utilizando a classe ColumnTransformer.

Essa abordagem garante que as transformações sejam aplicadas de forma consistente tanto nos dados de treinamento quanto nos dados de teste, reduzindo o risco de erros e vazamento de informações durante o treinamento dos modelos.

```python
# Separação das variáveis numéricas e categóricas

numericas = X.select_dtypes(include=["int64", "float64"]).columns.tolist()
categoricas = X.select_dtypes(include=["object"]).columns.tolist()

print("Variáveis numéricas:")
print(numericas)

print("\nVariáveis categóricas:")
print(categoricas)
```

```python
numeric_transformer = Pipeline(
    steps=[
        ("scaler", StandardScaler())
    ]
)
```

```python
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]
)
```

A estratégia most_frequent foi escolhida por se tratar de variáveis categóricas. Dessa forma, os valores ausentes são substituídos pela categoria mais recorrente, reduzindo o impacto da perda de informação e permitindo que todos os registros sejam utilizados no treinamento dos modelos.


```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numericas),
        ("cat", categorical_transformer, categoricas)
    ]
)

preprocessor
```

Nesta etapa, os tratamentos definidos anteriormente são reunidos em um único objeto de pré-processamento.

As variáveis numéricas passam pelo processo de padronização utilizando StandardScaler, que transforma os dados para uma escala com média próxima de zero e desvio padrão igual a um. Esse procedimento reduz diferenças de magnitude entre os atributos e contribui para um treinamento mais estável dos modelos.

As variáveis categóricas têm os valores ausentes tratados por imputação da categoria mais frequente e, posteriormente, são convertidas para formato numérico por meio do OneHotEncoder. Essa técnica cria representações binárias para cada categoria, permitindo que os algoritmos utilizem essas informações sem assumir relações de ordem entre os valores.

A utilização do ColumnTransformer garante que cada grupo de variáveis receba o tratamento adequado dentro de um fluxo único, reproduzível e consistente, assegurando que as mesmas transformações sejam aplicadas tanto aos dados de treinamento quanto aos dados de teste.

## 5.4 Divisão entre treino e teste

Antes da divisão dos dados, foi definido que todas as transformações de pré-processamento seriam incorporadas a um pipeline. Dessa forma, as etapas de imputação, codificação e padronização seriam aprendidas exclusivamente a partir dos dados de treinamento e posteriormente aplicadas aos dados de teste, evitando vazamento de informações (data leakage).

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=SEED
)

print("Treino:", X_train.shape)
print("Teste :", X_test.shape)
```


A estratégia de holdout foi adotada por ser adequada ao objetivo deste trabalho e ao tamanho do conjunto de dados disponível. Como a base contém aproximadamente 1 milhão de registros, a utilização de uma única divisão entre treinamento e teste é suficiente para obter estimativas confiáveis de desempenho, sem a necessidade de técnicas mais custosas computacionalmente, como validação cruzada.

Foi utilizada a proporção de 80% dos registros para treinamento e 20% para teste. Essa divisão permite disponibilizar uma quantidade expressiva de observações para o aprendizado dos modelos, ao mesmo tempo em que reserva um conjunto independente para avaliação da capacidade de generalização.

Por se tratar de um problema de regressão, não foi necessária a aplicação de estratificação, uma vez que não existem classes a serem balanceadas.

Além disso, para evitar vazamento de informações (data leakage), todas as etapas de pré-processamento foram incorporadas a pipelines. Dessa forma, os parâmetros de imputação, padronização e codificação são aprendidos exclusivamente a partir dos dados de treinamento e posteriormente aplicados aos dados de teste.

O conjunto de treinamento será utilizado para o ajuste dos modelos e aprendizado dos padrões presentes nos dados, enquanto o conjunto de teste permanecerá isolado durante o treinamento, sendo utilizado apenas na etapa final de avaliação.

Foi definido o parâmetro `random_state=SEED` para garantir a reprodutibilidade dos resultados, assegurando que a divisão dos dados possa ser reproduzida em futuras execuções do experimento.


### Considerações

As decisões de pré-processamento foram definidas com o objetivo de preparar os dados para a etapa de modelagem, preservando a maior quantidade possível de informações e evitando vazamento de dados.

As variáveis medical_history e family_medical_history apresentavam valores ausentes. Como se tratam de variáveis categóricas e aproximadamente 25% dos registros continham valores faltantes, optou-se pela imputação utilizando a categoria mais frequente (most frequent), evitando a perda de observações relevantes.

As variáveis numéricas age, bmi e children foram padronizadas utilizando StandardScaler, reduzindo diferenças de escala e favorecendo o treinamento dos modelos.

As variáveis categóricas foram transformadas por meio do OneHotEncoder, permitindo sua utilização pelos algoritmos de Machine Learning.

As variáveis bmi_categoria e faixa_idade, criadas durante a análise exploratória, não foram utilizadas na modelagem por derivarem diretamente das variáveis originais bmi e age, o que poderia introduzir redundância sem ganho informacional relevante.

Todas as transformações foram incorporadas a um pipeline de pré-processamento, garantindo reprodutibilidade e evitando vazamento de informações entre os conjuntos de treinamento e teste.

# 6. Modelagem

Nesta etapa serão treinados e avaliados diferentes algoritmos de regressão para prever o custo do seguro de saúde (charges). Foram selecionados modelos de complexidade crescente para permitir a comparação entre abordagens lineares e não lineares.

A seleção dos algoritmos buscou representar diferentes níveis de complexidade. A Regressão Linear foi utilizada como modelo de referência (baseline), a Árvore de Decisão como uma alternativa capaz de capturar relações não lineares e o Random Forest como um modelo baseado em ensemble, frequentemente utilizado em problemas tabulares devido à sua robustez e capacidade preditiva.

## 6.1 Modelo Baseline — Regressão Linear
A Regressão Linear foi utilizada como modelo baseline por ser uma técnica simples, amplamente conhecida e facilmente interpretável.

Os resultados obtidos servirão como referência para avaliar se modelos mais complexos conseguem gerar ganhos significativos de desempenho.

```python
pipeline_lr = Pipeline(
    steps=[
        ("preprocessamento", preprocessor),
        ("modelo", LinearRegression())
    ]
)
```

```python
pipeline_lr.fit(X_train, y_train)
```

```python
y_pred_lr = pipeline_lr.predict(X_test)
```

Após o treinamento do modelo, foram realizadas previsões sobre o conjunto de teste. Em seguida, o desempenho foi mensurado utilizando as métricas MAE (Mean Absolute Error), RMSE (Root Mean Squared Error) e R² (Coeficiente de Determinação).

```python
mae_lr = mean_absolute_error(y_test, y_pred_lr)
rmse_lr = root_mean_squared_error(y_test, y_pred_lr)
r2_lr = r2_score(y_test, y_pred_lr)

print("MAE :", mae_lr)
print("RMSE:", rmse_lr)
print("R²  :", r2_lr)
```

## 6.2 Árvore de Decisão

A Árvore de Decisão foi escolhida por sua capacidade de modelar relações não lineares entre as variáveis preditoras e a variável alvo.

Diferentemente da Regressão Linear, esse algoritmo não assume uma relação linear entre as variáveis, podendo capturar interações mais complexas presentes nos dados.

```python
pipeline_dt = Pipeline(
    steps=[
        ("preprocessamento", preprocessor),
        ("modelo", DecisionTreeRegressor(
            random_state=SEED
        ))
    ]
)

pipeline_dt.fit(X_train, y_train)

y_pred_dt = pipeline_dt.predict(X_test)
```

O modelo treinado foi aplicado ao conjunto de teste para geração das previsões. As métricas de desempenho foram calculadas para permitir posterior comparação com os demais algoritmos avaliados.

```python
mae_dt = mean_absolute_error(y_test, y_pred_dt)
rmse_dt = root_mean_squared_error(y_test, y_pred_dt)
r2_dt = r2_score(y_test, y_pred_dt)

print("MAE :", mae_dt)
print("RMSE:", rmse_dt)
print("R²  :", r2_dt)
```

## 6.3 Random Forest

O Random Forest consiste em um conjunto de árvores de decisão treinadas de forma independente. As previsões individuais são combinadas para produzir um resultado final mais robusto e menos suscetível ao sobreajuste (overfitting).

Esse algoritmo costuma apresentar bom desempenho em problemas tabulares e servirá como modelo mais sofisticado para comparação com os modelos anteriores.

```python
pipeline_rf = Pipeline(
    steps=[
        ("preprocessamento", preprocessor),
        ("modelo", RandomForestRegressor(
            n_estimators=100,
            random_state=SEED,
            n_jobs=-1
        ))
    ]
)

pipeline_rf.fit(X_train, y_train)

y_pred_rf = pipeline_rf.predict(X_test)
```

Após o treinamento do conjunto de árvores, foram geradas previsões para os dados de teste. As métricas calculadas serão utilizadas na seção seguinte para comparação do desempenho entre os modelos.

```python
mae_rf = mean_absolute_error(y_test, y_pred_rf)
rmse_rf = root_mean_squared_error(y_test, y_pred_rf)
r2_rf = r2_score(y_test, y_pred_rf)

print("MAE :", mae_rf)
print("RMSE:", rmse_rf)
print("R²  :", r2_rf)
```

#11. Conclusão

O presente trabalho teve como objetivo desenvolver e avaliar modelos de Machine Learning capazes de prever os custos de seguros de saúde a partir de características demográficas, comportamentais e clínicas dos segurados. Para isso, foi utilizado um conjunto de dados sintético contendo aproximadamente um milhão de registros, permitindo a aplicação de técnicas de regressão supervisionada em larga escala.

A análise exploratória dos dados possibilitou compreender as principais características da base e identificar fatores potencialmente relacionados ao custo do seguro. Os resultados indicaram que variáveis como tabagismo, nível de cobertura contratada e histórico médico apresentaram maior influência sobre os valores observados. Por outro lado, variáveis como idade, IMC e quantidade de dependentes demonstraram impacto mais moderado, embora ainda relevantes para o problema estudado. Também foi constatada a presença de valores ausentes em algumas variáveis categóricas, os quais foram tratados por meio de imputação utilizando a categoria mais frequente.

Na etapa de preparação dos dados, foram aplicadas boas práticas de Machine Learning, incluindo separação entre variáveis preditoras e variável alvo, divisão dos dados em conjuntos de treinamento e teste, padronização das variáveis numéricas, codificação das variáveis categóricas e utilização de pipelines para evitar vazamento de informações e garantir a reprodutibilidade dos experimentos.

Foram avaliados três algoritmos de regressão com diferentes níveis de complexidade: Regressão Linear, Árvore de Decisão e Random Forest. A comparação foi realizada por meio das métricas MAE, RMSE e R², permitindo analisar tanto a magnitude dos erros quanto a capacidade explicativa dos modelos. Os resultados mostraram que a Regressão Linear apresentou o melhor desempenho geral, alcançando R² de aproximadamente 0,83 e superando modelos mais complexos. A Árvore de Decisão apresentou a menor capacidade de generalização, enquanto o Random Forest obteve resultados competitivos, porém insuficientes para superar o modelo linear.

Embora a otimização tenha identificado uma configuração considerada superior durante a validação, os resultados obtidos no conjunto de teste permaneceram muito próximos aos observados no modelo original, sem ganhos significativos de desempenho. Além disso, a comparação entre os resultados de treinamento e teste não indicou sinais relevantes de overfitting ou underfitting, evidenciando boa capacidade de generalização dos modelos avaliados.

Um dos principais achados deste trabalho foi a constatação de que modelos mais complexos não necessariamente produzem melhores resultados. Embora algoritmos baseados em árvores sejam frequentemente superiores em problemas tabulares, neste conjunto de dados a Regressão Linear apresentou o melhor desempenho, sugerindo que a relação entre as variáveis explicativas e o custo do seguro possui um comportamento predominantemente linear.
